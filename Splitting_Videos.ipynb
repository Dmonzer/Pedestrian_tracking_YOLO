```python

import csv
import subprocess

def time_to_seconds(time_str):
    minutes, seconds = map(int, time_str.split(':'))
    return minutes * 60 + seconds

def split_video(input_video, output_video, start_time, end_time):
    command = [
        'ffmpeg',
        '-i', input_video,
        '-ss', str(start_time),
        '-to', str(end_time),
        '-c', 'copy',
        output_video
    ]
    subprocess.run(command)

def main():
    input_video = '/content/drive/MyDrive/0_497Project/preliminarydata/zoom0009.mov'
    csv_file = '/content/drive/MyDrive/0_497Project/preliminarydata/zoom9_times.csv'

    with open(csv_file, 'r') as file:
        reader = csv.reader(file)
        next(reader)  # Skip the header row
        for i, row in enumerate(reader, start=1):
            start_time = time_to_seconds(row[0].strip())
            end_time = time_to_seconds(row[1].strip())
            output_video = f'/content/drive/MyDrive/0_497Project/preliminarydata/splittingzoom9/output_video_{i}.mov'
            split_video(input_video, output_video, start_time, end_time)

if __name__ == "__main__":
    main()

```
