```python
from ultralytics import YOLO

# Load a model
model = YOLO('yolov8n-pose.pt')  # load an official model
#model = YOLO('path/to/best.pt')  # load a custom model

# Define path to video file
source = '/content/drive/MyDrive/0_497Project/preliminarydata/ZOOM0007.mov'

# Run inference on the source
results = model(source, stream=True)  # generator of Results objects
```

```python

# Below VideoWriter object will create
# a frame of above defined The output
# is stored in 'filename.avi' file.
import numpy as np
import cv2
from PIL import Image

i=0
for result in results:
  im_array = result.plot()
  im = Image.fromarray(im_array[..., ::-1])  # RGB PIL image
  if i==0:
    video_output = cv2.VideoWriter('/content/drive/MyDrive/0_497Project/preliminarydata/results/ZOOM0007_with_pose.mp4', cv2.VideoWriter_fourcc(*'MJPG'),30, im.size)
  video_output.write(cv2.cvtColor(np.array(im), cv2.COLOR_RGB2BGR))
  #im.save(str(i)+'results.jpg')  # save image
  i+=1
```

```python
import cv2

# Load the video file
video_path = '/content/drive/MyDrive/0_497Project/preliminarydata/results/ZOOM0007_with_pose.mov'
cap = cv2.VideoCapture(video_path)

# Check if the video file is opened successfully
if not cap.isOpened():
    print("Error: Could not open the video file.")
else:
    # Read and display the video frame by frame
    while cap.isOpened():
        ret, frame = cap.read()
        if ret:
            cv2.imshow('Video', frame)

            # Press 'q' to quit the video playback
            if cv2.waitKey(25) & 0xFF == ord('q'):
                break
        else:
            break

    # Release the video capture object and close all windows
    cap.release()
    cv2.destroyAllWindows()
```
